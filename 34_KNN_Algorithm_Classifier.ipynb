# Classifier

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
# sklearn library ko dataset module bataw make_classification leko.
x,y  = make_classification(
    n_samples= 1000, # 1000 otaw row vaako sample banyo
    n_features= 3, #  total independent features i.e columns
    n_classes= 2, # It means multiclassification, 0 and 1. i.e binary target/label
    n_redundant= 1, # A column, B column ko combination banyeraw aako value C ma. i,e = A 10, C = 20 , C = A + B

)
# X = Independet, y = dependent
# Independent means = Input, dependent means = output
```

```python
y.shape
```

```python
# train model
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.33, random_state = 42)
```

```python
# Import KNeighborsClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
classifier = KNeighborsClassifier()
```

```python
classifier
```

```python
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
```

```python
# model predicted data
y_pred
```

```python
# original data
y_test
```

```python
# for classifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(y_pred, y_test))
print(accuracy_score(y_pred, y_test))       # output = 0.97 = It means 1000 ota maa 970 data predict gardey ko xa.It is good.
print(classification_report(y_pred, y_test))
```

```python
## GridSearchCV
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings('ignore')
```

```python
parameters = {
  'n_neighbors' : list(range(30)),  # which tests K values from 1 to 29.
  'p' : [1,2]                          # p determines the distance calculation used by KNN.
                                       #p = 1 → Manhattan distance.
                                       #p = 2 → Euclidean distance.
}
                                    # For KNN, n_neighbors means K, the number of neighbors used to make a prediction.
```

```python
clf = GridSearchCV(classifier, param_grid = parameters)
```

```python
## Splitting of train data to validation data
clf.fit(x_train, y_train)
```

```python
clf.best_params_
```

```python
classifier = KNeighborsClassifier(n_neighbors =  6, p = 2)
```

```python
classifier.fit(x_train, y_train)
```

```python
## Prediction
y_pred = classifier.predict(x_test)  # y_pred => model ley predict garey ko data. x_test => original data
print(confusion_matrix(y_pred, y_test))
print(accuracy_score(y_pred, y_test))
print(classification_report(y_pred, y_test))
```

```python
# Import pickle for saving and loading the model
import pickle
```

```python
# Filepath for saving the model
knn_file = 'knn.plk'
```

```python
# Save the trained model to a pickle file
with open(knn_file, 'wb') as file:
  pickle.dump(classifier,file)
```

```python
print(f'model saved to {knn_file}')
```

```python
# Load the model from the pickle file
with open(knn_file, 'rb') as file:
  loaded_model = pickle.load(file)
```

```python
# Test the loaded model with predictions
y_pred_loaded = loaded_model.predict(x_test)
print(y_pred_loaded)
```
