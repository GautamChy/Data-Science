```python
# Initialize the PorterStemmer
text = 'You have been selected as our lucky winner of a FREE cash prize worth £5,000. Claim your exclusive reward today by clicking the link below. This limited-time offer is available only to selected customers. Do not miss this amazing opportunity to receive free gifts and special bonuses. Click now to verify your account and collect your reward before the offer expires. Hurry up, your prize is waiting for you!'
ps = PorterStemmer()

def transform_text(text):
    text = text.lower()  # Convert to lowercase
    text = nltk.word_tokenize(text)  # Tokenize the text make a list of words

    y = []
    for i in text:
        if i.isalnum(): # Removing special characters
            y.append(i)

    text = y[:]
    y.clear()

    for i in text:
        if i not in stopwords.words('english') and i not in string.punctuation: # Remove stopwords and punctuation
            y.append(i)

    text = y[:]
    y.clear()

    for i in text:  # Apply stemming
        y.append(ps.stem(i))

    return " ".join(y)
```
