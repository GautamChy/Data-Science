# Stemming
* Stemming means cutting words to their root form.
* It removes suffiexes like ing,ed,etc.
* The roots(stem) may not be real word.
* Used in search engines,NLP tasks to match similar meaning.

# Lemmatization
* Lemmatization changes a word to its base dictionary form(root word)
* It uses grammer rules and vocabulary to find correct base form.
* The result is always a real word,unlike stemming.
* Used in making chatbots and language models.

Use lemmatization when
* Accuracy matters more than speed.
* Building advanced NLP systems using deep learning.

Use stemming when
* Speed is important.
* You don't need perfect word accuracy.
* When working on less memory/CPU device.

# TF(Term Frequency)

* S1 --> good boy
* S2 --> good girl
* S3 -->boy girl good

* TF = No. of repetition of words in sentence / No. of words in sentence.
* Formula : No. of repetition of words in sentence divided by No. of words in sentence.

* IDF = loge(No. of sentences / No. of sentences containing that word).
* Formula: loge(No. of sentences divided by No. of sentences containing that word.

SMS project

`````python
import pandas as pd
messages = pd.read_csv('sms.csv', index_col=0)
`````

`````python
messages
`````

`````python
import re
import nltk
nltk.download('all')
`````

`````python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
`````

`````python
len(messages)
`````

`````python
# This code performs text cleaning → lowercasing → tokenization → stopword removal → lemmatization → storage in a corpus.
# These steps prepare text for further NLP analysis or machine learning.
corpus = []
for i in range(0,len(messages)):
  review = re.sub('^a-zA-Z',' ',messages['text'][i]) # re.sub() is a function from Python's re module that replaces text matching a regular expression.' ', is the replacement: a space.
  review = review.lower() # Converts every letter to lowercase so that words with different capitalization are treated consistently.
  review = review.split() # The split() function divides a string into a list of words, using whitespace as the default separator.
  review = [lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)

`````

`````python
corpus
`````

`````python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=100)
x= tfidf.fit_transform(corpus).toarray()
`````
