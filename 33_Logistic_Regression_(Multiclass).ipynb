```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import pandas as pd
import numpy as np
```

```python
# Load the Iris dataset
iris = load_iris()
```

```python
iris
```

```python
print(iris.DESCR)
```

```python
# iris keys
iris.keys()
```

```python
# keys value i.e Data
iris.data
```

```python
iris.feature_names
```

```python
df = pd.DataFrame(iris.data, columns = iris.feature_names)
```

```python
df.head()
```

```python
# create new target column and put  dictionary target value
df['target'] = iris.target
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
# showing multiclass i.e 0,1,2. If 1,2 then it's called binary class.
iris.target
```

```python
# Independent
x = df.iloc[:,:-1]
```

```python
# Dependent
y = df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
# split dataset into training and test sets
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state = 42)
# 30% data for testing use and 70% data for training use.
#x_train, x_test = It is just a naming i.e small x or capital X ,we can write anything.
```

```python
x_train
```

```python
x_train.shape
```

```python
x_test
```

```python
x_test.shape
```

```python
# Multiclass logistic regression model
model = LogisticRegression()
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
# prediction
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
# real or original data. # without predict data
y_test
```

```python
# Evaluation metrics
conf_matrix = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print('Confusion Matrix:\n', conf_matrix)
print('\nAccuracy:', accuracy)
print('\nClassification Report:\n', report)
```

```python
# Import pickle for saving and loading the model
import pickle
```

```python
# Filepath for saving the model
pickle_file = 'logistic_regression_multiclass.plk'
```

```python
# Save the trained model to a pickle file
with open(pickle_file, 'wb') as file:
  pickle.dump(model,file)
```

```python
print(f'model saved to {pickle_file}')
```

```python
# Load the model from the pickle file
with open(pickle_file, 'rb') as file:
  loaded_model = pickle.load(file)
```

```python
# Test the loaded model with predictions
y_pred_loaded = loaded_model.predict(x_test)
print(y_pred_loaded)
```
