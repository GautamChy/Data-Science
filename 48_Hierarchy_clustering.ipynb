```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
## Import IRIS dataset
iris = datasets.load_iris()
```

```python
iris
```

```python
df = pd.DataFrame(iris.data)
```

```python
df.head()
```

```python
df.columns = iris.feature_names
```

```python
df.head()
```

```python
# Standardization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
X_scaled
```

```python
## Apply PCA
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2) # It reduce columns i.e before 4 column after apply PCS now it's 2 column
```

```python
pca_scaled = pca.fit_transform(X_scaled)
```

```python
pca_scaled
```

```python
plt.scatter(pca_scaled[:,0], pca_scaled[:,1], c=iris.target)
```

```python
## Apply Agglomerative Clustering
# To construct a dendrogram, we will use the scipy library.
import scipy.cluster.hierarchy as sc

# plot dendogram
plt.figure(figsize=(10, 7))
plt.title("Dendograms")

# Create dendogram
sc.dendrogram(sc.linkage(pca_scaled, method='ward'))
plt.title('Dendogram')
plt.xlabel('Sample Index')
plt.ylabel('Euclidean Distances')
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=4, metric='euclidean', linkage='ward')
cluster.fit(pca_scaled)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0], pca_scaled[:,1], c=cluster.labels_) # n_clusters=4 so, 4 different types of cluster color.
```
