```python
# making dependent features
X = df.drop('total_bill', axis = 1)
y = df['total_bill']
```

```python
# Dependent features
y.head()
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X,y, test_size = 0.25,random_state = 20)
```

```python
x_train.head()
```

# Feature Encoding(Label Encoding and One Hot Encoding)
* Convert above column value into Numerical data
i.e Male 1

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
# fit ( only do in training)
x_train['sex'] = le1.fit_transform(x_train['sex'])
x_train['smoker'] = le2.fit_transform(x_train['smoker'])
x_train['time'] = le3.fit_transform(x_train['time'])
```

```python
x_train.head()
```

```python
# Nover do fit in testing
x_test['sex'] = le1.transform(x_test['sex'])
x_test['smoker'] = le2.transform(x_test['smoker'])
x_test['time'] = le3.transform(x_test['time'])
```

```python
## One Hot Encoding -- Column Transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers = [('onehot', OneHotEncoder(drop = 'first'), [3])], remainder = 'passthrough')
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
x_train = ct.fit_transform(x_train)
x_train
```

```python
x_test = ct.transform(x_test)
x_test
```

```python
## SVR Model
from sklearn.svm import SVR
model = SVR()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print(r2_score(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
```

```python
# Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV
# defined parameter range
param_grid = {'C': [0.1, 1, 10, 100, 1000], 'gamma': [1, 0.1, 0.01, 0.001, 0.0001]}
```

```python
grid = GridSearchCV(SVR(), param_grid, refit = True, verbose = 3)

# fitting the model for grid search
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
grid_predictions = grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print(r2_score(y_test, grid_predictions))
print(mean_absolute_error(y_test, grid_predictions))
print(mean_squared_error(y_test, grid_predictions))
```
