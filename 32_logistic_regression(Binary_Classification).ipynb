# Binary Classification

#  Categorical Variables and Continuous Variables

## 1. Categorical Variable
# NOTE: Output ma categorical variable vaako bela LOGISTIC REGRESSION use hunxa

# NOTE : Categorical Variable mainly used in Logistic Regression

A **categorical variable** puts observations into **groups or categories**. The values represent types or labels rather than amounts.

### Examples:
- Gender: Male / Female / Other
- Blood group: A / B / AB / O
- Smoker: Yes / No
- Education: School / College / University
- Eye colour: Brown / Blue / Green

👉 **Think:** "Which group?"

---

## Simple Example

Suppose we study whether a person has a disease:

| Variable | Type | Example |
|---|---|---|
| Age | Continuous | 45.5 years |
| Weight | Continuous | 72.3 kg |
| Blood pressure | Continuous | 125.6 |
| Gender | Categorical | Male |
| Smoker | Categorical | Yes |
| Disease | Categorical | Yes/No |

---

## Variables in Logistic Regression

In **logistic regression**, the **dependent variable** is commonly categorical, usually **Yes/No or 0/1**.

The **independent variables** can be:

- Continuous variables
- Categorical variables
- Or a mixture of both

### Example:

**Dependent variable (Y):**
- Disease → Yes/No

**Independent variables (X):**
- Age → Continuous
- Weight → Continuous
- Gender → Categorical
- Smoker → Categorical



## 2. Continuous Variable

A **continuous variable** is a value that can take many possible numerical values, including decimals.

### Examples:
- Height: 170.5 cm
- Weight: 68.7 kg
- Age: 25.5 years
- Temperature: 36.8°C
- Income: £2,450.75

👉 **Think:** "How much?" or "How many?"

### In simple words:

> Logistic regression can use continuous and categorical independent variables to predict a categorical outcome.

#  Dependent and Independent

```python
# Independent
x = df_copy.iloc[:,:-1]
```

```python
# Dependent
y = df_copy.iloc[:,-1]
```

```python
y
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size = 0.33, random_state = 121)
```

```python
classifier = LogisticRegression()
```

```python
## prediction
y_pred = classifier.predict(X_test)
```

# Confusion matrix, accuracy score, classification report

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(y_test, y_pred))  # y_test is original and y_predict is after prediction
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

# Hyperparameter Tuning
- Hyperparameter tuning = trying different model settings to find the settings that give the best performance.

```python
# GridSearchCV is a tool in Python's scikit-learn that helps you find the best hyperparameters for a machine learning model.
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings('ignore')
```
