```python
# Create columnTranform with 3 types of transformers
num_features = x.select_dtypes(exclude='object').columns
onehot_columns = ['seller_type', 'fuel_type', 'transmission_type']

from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

numer_transformer = StandardScaler()
oh_transformer = OneHotEncoder(drop='first')

preprocessor = ColumnTransformer(
    [
        ('OneHotEncoder', oh_transformer, onehot_columns),
        ('StandardScaler', numer_transformer, num_features)
    ], remainder = 'passthrough'
)
```

```python
# Separa dataset into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

Model Training and Model Selection

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# Create a function to evaluate the model.
def evaluate_model(true, predicted):# evaluate_model is function name,true and predicted is parameter
    mae = mean_absolute_error(true, predicted)
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(true, predicted)
    return mae, rmse, r2, mse
    # true = It is actual value, predicted = It is predicted by model
```

```python
## Begining of model training
models = {
    'GradientBoostingRegressor': GradientBoostingRegressor(),
    'Decision Tree': DecisionTreeRegressor()
}

for i in range(len(list(models))):
  model = list(models.values())[i]
  model.fit(X_train, y_train) # Train the model

  # Make predictions
  y_test_pred = model.predict(X_test)
  y_train_pred = model.predict(X_train)

  # Evaluate the model
  mae_train, rmse_train, r2_train, mse_train = evaluate_model(y_train, y_train_pred)
  mae_test, rmse_test, r2_test, mse_train = evaluate_model(y_test, y_test_pred )

  print(f'Model: {list(models.keys())[i]}')

  print("Model Performance for Training Set")
  print(" Root Mean Squared Error: {: 4f}".format(rmse_train))
  print(" Mean Absolute Error: {: 4f}".format(mae_train))
  print(" R2 Score: {: 4f}".format(r2_train))

  print("................................")

  print("Model Performaance on Testing Set")
  print("Root Mean Squared Error: {:4f}".format(rmse_test))
  print("Mean Absolute Error: {: 4f}".format(mae_test))
  print("R2 Score: {: 4f}".format(r2_test))

  print('=' *35)
  print('\n')
```

```python
# Initialize few parameters for hyperparameter tuning
gb_params = {
    'loss': ['squared_error', 'absolute_error', 'huber'],
    'criterion': ['friedman_mse', 'squared_error', 'mse'],
    'min_samples_split': [2, 8, 15, 20],
    'n_estimators' :[50, 100, 150, 200, 250],
    'max_depth' : [3,5,8,10,None],
    'learning_rate': [0.01, 0.05, 0.1, 0.2, 0.3]
}
```

```python
# model list for hyperparameter tuning
randomcv_models = [
    ('GradientBoostingRegressor', GradientBoostingRegressor(), gb_params)
]
```

```python
# Hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV

model_params = {}

for name, model, params in randomcv_models:
  random = RandomizedSearchCV(estimator=model,param_distributions=params,n_iter=100, cv=3,verbose=2,n_jobs=-1)
  random.fit(X_train, y_train)
  model_params[name] = random.best_params_

for model_name in model_params:
  print(f'........Best Parameters for {model_name}.........')
  print(model_params[model_name])
```
